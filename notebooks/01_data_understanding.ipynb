```python
# Convert date columns into datetime format
batches_df["received_at"] = pd.to_datetime(batches_df["received_at"])
batches_df["snapshot_date"] = pd.to_datetime(batches_df["snapshot_date"])
batches_df["expiry_date"] = pd.to_datetime(batches_df["expiry_date"])

# Calculate actual days remaining
batches_df["calculated_days_to_expiry"] = (
    batches_df["expiry_date"] - batches_df["snapshot_date"]
).dt.days

# Compare calculated and existing values
batches_df["expiry_days_mismatch"] = (
    batches_df["days_to_expiry"] !=
    batches_df["calculated_days_to_expiry"]
)

print("Expiry date mismatches:",
      batches_df["expiry_days_mismatch"].sum())

print("\nExpired batches:")
print((batches_df["calculated_days_to_expiry"] < 0).sum())

print("\nBatches expiring today:")
print((batches_df["calculated_days_to_expiry"] == 0).sum())
```

```python
print("Remaining quantity greater than initial quantity:",
      (batches_df["quantity_remaining"] >
       batches_df["initial_quantity"]).sum())

print("Negative initial quantity:",
      (batches_df["initial_quantity"] < 0).sum())

print("Negative remaining quantity:",
      (batches_df["quantity_remaining"] < 0).sum())

print("\nInventory quantity summary:")

display(
    batches_df[
        ["initial_quantity", "quantity_remaining"]
    ].describe()
)
```

```python
# Merge product master with batch information
batch_price_check = batches_df.merge(
    products_df[["product_id", "selling_price", "unit_cost"]],
    on="product_id",
    how="left",
    suffixes=("_batch", "_product")
)

print("Selling price mismatches:",
      (batch_price_check["unit_price"] !=
       batch_price_check["selling_price"]).sum())

print("Unit cost mismatches:",
      (batch_price_check["unit_cost_batch"] !=
       batch_price_check["unit_cost_product"]).sum())
```

```python
print("Total discount scenarios:", len(discount_df))

print("\nUnique batches in discount dataset:",
      discount_df["batch_id"].nunique())

print("\nDiscount percentages:")
display(
    discount_df["discount_pct"].value_counts().sort_index()
)

print("\nDuplicate batch and discount combinations:",
      discount_df.duplicated(
          subset=["batch_id", "discount_pct"]
      ).sum())

print("\nDiscount percentage outside 0-100:",
      (~discount_df["discount_pct"].between(0, 100)).sum())

print("\nEstimated units sold greater than available quantity:",
      (discount_df["estimated_units_sold_simulated"] >
       discount_df["quantity_remaining"]).sum())
```

```python
print("Expiry Risk Label vs Days to Expiry")

display(
    batches_df.groupby("expiry_risk_label").agg(
        total_batches=("batch_id", "count"),
        average_days_to_expiry=("days_to_expiry", "mean"),
        average_remaining_quantity=("quantity_remaining", "mean"),
        average_expected_demand=("expected_demand_before_expiry", "mean"),
        average_simulated_waste=("waste_units_actual_simulated", "mean")
    ).round(2)
)

print("\nExpiry risk label vs expired status")

display(
    pd.crosstab(
        batches_df["expiry_risk_label"],
        batches_df["days_to_expiry"] < 0,
        margins=True
    )
)
```

```python
# Find expired batches labelled as low risk
expired_low_risk = batches_df[
    (batches_df["days_to_expiry"] < 0) &
    (batches_df["expiry_risk_label"] == 0)
]

print("Expired batches with label 0:", len(expired_low_risk))

display(
    expired_low_risk[
        [
            "batch_id",
            "product_id",
            "snapshot_date",
            "expiry_date",
            "days_to_expiry",
            "quantity_remaining",
            "avg_daily_sales_7d",
            "expected_demand_before_expiry",
            "waste_units_actual_simulated",
            "expiry_risk_label"
        ]
    ].head(15)
)
```

```python
batches_df["is_expired"] = batches_df["days_to_expiry"] < 0

expiry_comparison = batches_df.groupby("is_expired").agg(
    total_batches=("batch_id", "count"),
    average_remaining_stock=("quantity_remaining", "mean"),
    average_daily_sales=("avg_daily_sales_7d", "mean"),
    average_expected_demand=("expected_demand_before_expiry", "mean"),
    average_simulated_waste=("waste_units_actual_simulated", "mean"),
    high_risk_batches=("expiry_risk_label", "sum")
).round(2)

display(expiry_comparison)
```

```python
batches_df["simple_demand_estimate"] = (
    batches_df["avg_daily_sales_7d"] *
    batches_df["days_to_expiry"].clip(lower=0)
)

batches_df["demand_difference"] = (
    batches_df["expected_demand_before_expiry"] -
    batches_df["simple_demand_estimate"]
)

display(
    batches_df[
        [
            "batch_id",
            "days_to_expiry",
            "avg_daily_sales_7d",
            "expected_demand_before_expiry",
            "simple_demand_estimate",
            "demand_difference",
            "expiry_risk_label"
        ]
    ].head(20)
)

print(
    "Average absolute demand difference:",
    batches_df["demand_difference"].abs().mean().round(2)
)
```

```python
print("Average feature values by expiry risk label:")

display(
    batches_df.groupby("expiry_risk_label")[
        [
            "days_to_expiry",
            "quantity_remaining",
            "avg_daily_sales_7d",
            "expected_demand_before_expiry",
            "discount_pct",
            "waste_units_actual_simulated"
        ]
    ].mean().round(2)
)

print("\nCorrelation with expiry risk label:")

display(
    batches_df[
        [
            "days_to_expiry",
            "quantity_remaining",
            "avg_daily_sales_7d",
            "expected_demand_before_expiry",
            "discount_pct",
            "waste_units_actual_simulated",
            "expiry_risk_label"
        ]
    ].corr()["expiry_risk_label"].sort_values(ascending=False)
)
```

```python
batches_df["inventory_surplus"] = (
    batches_df["quantity_remaining"] -
    batches_df["expected_demand_before_expiry"]
)

print("Average inventory surplus by risk label:")

display(
    batches_df.groupby("expiry_risk_label")[
        ["inventory_surplus"]
    ].agg(["mean", "min", "max"])
)
```

```python
batches_df["predicted_risk_rule"] = (
    batches_df["quantity_remaining"] >
    batches_df["expected_demand_before_expiry"]
).astype(int)

print("Original labels vs candidate rule:")

display(
    pd.crosstab(
        batches_df["expiry_risk_label"],
        batches_df["predicted_risk_rule"],
        rownames=["Original Label"],
        colnames=["Candidate Rule"]
    )
)

print(
    "Matching records:",
    (
        batches_df["expiry_risk_label"] ==
        batches_df["predicted_risk_rule"]
    ).sum()
)

print(
    "Mismatching records:",
    (
        batches_df["expiry_risk_label"] !=
        batches_df["predicted_risk_rule"]
    ).sum()
)
```

```python
mismatched_batches = batches_df[
    batches_df["expiry_risk_label"] !=
    batches_df["predicted_risk_rule"]
]

print("Total mismatched batches:", len(mismatched_batches))

display(
    mismatched_batches[
        [
            "batch_id",
            "days_to_expiry",
            "quantity_remaining",
            "avg_daily_sales_7d",
            "expected_demand_before_expiry",
            "inventory_surplus",
            "waste_units_actual_simulated",
            "expiry_risk_label",
            "predicted_risk_rule"
        ]
    ].head(20)
)
```

```python
print("Risk distribution for non-expired batches:")

display(
    batches_df[batches_df["days_to_expiry"] >= 0][
        "expiry_risk_label"
    ].value_counts()
)

print("\nRisk distribution for expired batches:")

display(
    batches_df[batches_df["days_to_expiry"] < 0][
        "expiry_risk_label"
    ].value_counts()
)
```

```python
# Compare original label with candidate rule
batches_df["comparison_group"] = "Match"

batches_df.loc[
    (batches_df["expiry_risk_label"] == 1) &
    (batches_df["predicted_risk_rule"] == 0),
    "comparison_group"
] = "Original High Risk - Rule Low Risk"

batches_df.loc[
    (batches_df["expiry_risk_label"] == 0) &
    (batches_df["predicted_risk_rule"] == 1),
    "comparison_group"
] = "Original Low Risk - Rule High Risk"

print(batches_df["comparison_group"].value_counts())
```

```python
comparison_summary = batches_df.groupby("comparison_group").agg(
    total_batches=("batch_id", "count"),
    avg_days_to_expiry=("days_to_expiry", "mean"),
    avg_remaining_stock=("quantity_remaining", "mean"),
    avg_daily_sales=("avg_daily_sales_7d", "mean"),
    avg_expected_demand=("expected_demand_before_expiry", "mean"),
    avg_simulated_waste=("waste_units_actual_simulated", "mean")
).round(2)

display(comparison_summary)
```

```python
print("Original risk label distribution:")

display(
    batches_df.groupby(
        ["is_expired", "expiry_risk_label"]
    ).size().unstack(fill_value=0)
)

print("\nOriginal label distribution by comparison group:")

display(
    pd.crosstab(
        batches_df["comparison_group"],
        batches_df["expiry_risk_label"]
    )
)
```

```python
batches_df["has_simulated_waste"] = (
    batches_df["waste_units_actual_simulated"] > 0
).astype(int)

display(
    pd.crosstab(
        batches_df["expiry_risk_label"],
        batches_df["has_simulated_waste"],
        margins=True
    )
)

print(
    "Labels matching waste-based rule:",
    (
        batches_df["expiry_risk_label"] ==
        batches_df["has_simulated_waste"]
    ).sum()
)

print(
    "Labels not matching waste-based rule:",
    (
        batches_df["expiry_risk_label"] !=
        batches_df["has_simulated_waste"]
    ).sum()
)
```

```python
display(
    batches_df.groupby("expiry_risk_label").agg(
        total_batches=("batch_id", "count"),
        zero_waste_batches=(
            "waste_units_actual_simulated",
            lambda x: (x == 0).sum()
        ),
        positive_waste_batches=(
            "waste_units_actual_simulated",
            lambda x: (x > 0).sum()
        ),
        average_waste=(
            "waste_units_actual_simulated",
            "mean"
        ),
        maximum_waste=(
            "waste_units_actual_simulated",
            "max"
        )
    )
)
```

```python
waste_correlation = batches_df[
    [
        "days_to_expiry",
        "quantity_remaining",
        "avg_daily_sales_7d",
        "expected_demand_before_expiry",
        "discount_pct",
        "unit_price",
        "unit_cost",
        "initial_quantity",
        "waste_units_actual_simulated"
    ]
].corr()

display(
    waste_correlation[
        ["waste_units_actual_simulated"]
    ].sort_values(
        by="waste_units_actual_simulated",
        ascending=False
    )
)
```

```python
display(
    batches_df.groupby("expiry_risk_label").agg(
        average_remaining_stock=("quantity_remaining", "mean"),
        average_initial_stock=("initial_quantity", "mean"),
        average_daily_sales=("avg_daily_sales_7d", "mean"),
        average_days_to_expiry=("days_to_expiry", "mean"),
        average_waste=("waste_units_actual_simulated", "mean")
    ).round(2)
)
```
